```python
from faker import Faker

# Crear instancia de Faker
fake = Faker('es_MX')

# Generar 100 nombres de pila aleatorios
nombres_aleatorios = [fake.first_name() for _ in range(100)]

nombres_aleatorios = [nombre.lower() for nombre in nombres_aleatorios]

# Mostrar los nombres
for nombre in nombres_aleatorios:
    print(nombre)

```

```python
b = {}
for w in nombres_aleatorios:
  chs = ['<S>'] + list(w) + ['<E>']
  for ch1, ch2 in zip(chs, chs[1:]):
    bigram = (ch1,ch2)
    b[bigram] = b.get(bigram, 0) + 1
    print(ch1,ch2)
```

```python
import numpy as np

N = np.zeros((32, 32), dtype=int)
```

```python
chars = sorted(list(set(''.join(nombres_aleatorios))))
len(chars)
```

```python
chars = sorted(list(set(''.join(nombres_aleatorios))))
stoi = {s:i+1 for i,s in enumerate(chars)}
stoi['<S>'] = 0
stoi['<E>'] = 31
for w in nombres_aleatorios:
  chs = ['<S>'] + list(w) + ['<E>']
  for ch1, ch2 in zip(chs, chs[1:]):
    bigram = (ch1,ch2)
    b[bigram] = b.get(bigram, 0) + 1
    ix1 = stoi[ch1]
    ix2 = stoi[ch2]
    N[ix1, ix2] += 1
itos = {i:s for s,i in stoi.items()}
```

```python
chs = ['.'] + list(w) + ['.']
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.figure(figsize = (16,16))
plt.imshow(N, cmap = 'Blues')
for i in range(27):
  for j in range(27):
    chstr = itos[i] + itos[j]
    plt.text(j, i, chstr, ha = 'center', va = 'bottom', color = 'gray')
    plt.text(j, i, N[i,j].item(), va = 'top', color = 'gray')
plt.axis('off');
```
